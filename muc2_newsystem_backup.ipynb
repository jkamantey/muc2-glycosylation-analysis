# Creating the graph for muc2_ambernew_system

- Plotting the rmsd from 0 to 1.2 us
- Plotting the rmsf from 200 to 1.2us


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-darkgrid')
```

## File Location

```python
# Time step
dt_ns = 0.12018  # ns per frame

# Output directory
output_dir = "/home/computer2/analysis_simulations/analysis/plots"
os.makedirs(output_dir, exist_ok=True)

print("Output directory:", output_dir)
```

## Load Data - All 3 Systems

```python
# Load RMSF datasets for all 3 systems
df_rmsf_1 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_ambernew_system/rmsf.dat",
                         sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_rmsf_2 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_ambernew_system2/rmsf_2.dat",
                         sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_rmsf_3 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_ambernew_system3/rmsf_3.dat",
                         sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])

# Load RMSD datasets for all 3 systems
df_rmsd_1 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_ambernew_system/rmsd_beginning.dat",
                          sep=r"\s+", skiprows=1, names=["Frame", "RMSD"])
df_rmsd_2 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_ambernew_system2/rmsd_beginning_2.dat",
                          sep=r"\s+", skiprows=1, names=["Frame", "RMSD"])
df_rmsd_3 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_ambernew_system3/rmsd_beginning3.dat",
                          sep=r"\s+", skiprows=1, names=["Frame", "RMSD"])

# Convert frames to time (ns)
time_ns_1 = (df_rmsd_1["Frame"] - df_rmsd_1["Frame"].iloc[0]) * dt_ns
time_ns_2 = (df_rmsd_2["Frame"] - df_rmsd_2["Frame"].iloc[0]) * dt_ns
time_ns_3 = (df_rmsd_3["Frame"] - df_rmsd_3["Frame"].iloc[0]) * dt_ns

print("All 3 RMSF and RMSD datasets loaded")
print(f"Run 1 - RMSF: {df_rmsf_1.shape}, RMSD: {df_rmsd_1.shape}")
print(f"Run 2 - RMSF: {df_rmsf_2.shape}, RMSD: {df_rmsd_2.shape}")
print(f"Run 3 - RMSF: {df_rmsf_3.shape}, RMSD: {df_rmsd_3.shape}")
```

## RMSD - Individual Plots

```python
# RMSD - Run 1
fig, ax = plt.subplots(figsize=(14, 7), dpi=100)
ax.plot(time_ns_1, df_rmsd_1["RMSD"], linewidth=2.5, color="#2E86AB", alpha=0.85,
        marker="o", markersize=3, markevery=max(1, len(df_rmsd_1)//50), label="Run 1")
ax.fill_between(time_ns_1, df_rmsd_1["RMSD"], alpha=0.1, color="#2E86AB")
ax.set_xlabel("Time (ns)", fontsize=13, fontweight="bold")
ax.set_ylabel("RMSD (\u00c5)", fontsize=13, fontweight="bold")
ax.set_title("RMSD - Run 1", fontsize=15, fontweight="bold", pad=20)
ax.grid(True, alpha=0.3, linestyle="--", linewidth=0.5)
ax.set_facecolor("white")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(True)
ax.spines["left"].set_color("black")
ax.spines["left"].set_linewidth(1.5)
ax.spines["bottom"].set_visible(True)
ax.spines["bottom"].set_color("black")
ax.spines["bottom"].set_linewidth(1.5)
ax.tick_params(axis="both", which="major", labelsize=11, length=6, width=1.5, direction="out", colors="black")
plt.tight_layout()
out = os.path.join(output_dir, "rmsd_ambernew_system1.png")
plt.savefig(out, dpi=800, bbox_inches="tight", facecolor="white")
plt.show()
print(f"Saved: {out}")
```

## RMSF Fluctuation Comparison - MUC2 Glycans vs No Glycans

```python
# Load noglycans RMSF data (3 runs)
df_ng1 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans/rmsf.dat",
                      sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_ng2 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans1/rmsf_noglyc_1.dat",
                      sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_ng3 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans3/rmsf3_noglyc.dat",
                      sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])

# Compute mean ± std for muc2_newsystem (glycans)
rmsf_new = pd.concat([
    df_rmsf_1.set_index("Residue")["RMSF"],
    df_rmsf_2.set_index("Residue")["RMSF"],
    df_rmsf_3.set_index("Residue")["RMSF"]
], axis=1, keys=["Run1", "Run2", "Run3"])
mean_new = rmsf_new.mean(axis=1)
std_new  = rmsf_new.std(axis=1)
residues_new = mean_new.index

# Compute mean ± std for muc2_noglycans
rmsf_nog = pd.concat([
    df_ng1.set_index("Residue")["RMSF"],
    df_ng2.set_index("Residue")["RMSF"],
    df_ng3.set_index("Residue")["RMSF"]
], axis=1, keys=["Run1", "Run2", "Run3"])
mean_nog = rmsf_nog.mean(axis=1)
std_nog  = rmsf_nog.std(axis=1)
residues_nog = mean_nog.index

# Plot both systems on same graph
fig, ax = plt.subplots(figsize=(14, 7), dpi=100)

ax.plot(residues_new, mean_new, linewidth=2.5, color="#2E86AB", label="MUC2 Glycans (mean)")
ax.fill_between(residues_new, mean_new - std_new, mean_new + std_new,
                alpha=0.25, color="#2E86AB", label="MUC2 Glycans ± Std")

ax.plot(residues_nog, mean_nog, linewidth=2.5, color="#A23B72", label="MUC2 No Glycans (mean)")
ax.fill_between(residues_nog, mean_nog - std_nog, mean_nog + std_nog,
                alpha=0.25, color="#A23B72", label="MUC2 No Glycans ± Std")

ax.set_xlabel("Residue Number", fontsize=13, fontweight="bold")
ax.set_ylabel("RMSF (Å)", fontsize=13, fontweight="bold")
ax.set_title("RMSF Fluctuation Comparison - Glycans vs No Glycans", fontsize=15, fontweight="bold", pad=20)

max_res = max(residues_new.max(), residues_nog.max())
ax.set_xticks(np.arange(0, max_res + 100, 100))
ax.set_xlim(left=min(residues_new.min(), residues_nog.min()))
ax.set_ylim(bottom=0)

ax.grid(True, alpha=0.3, linestyle="--", linewidth=0.5)
ax.set_facecolor("white")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(True)
ax.spines["left"].set_color("black")
ax.spines["left"].set_linewidth(1.5)
ax.spines["bottom"].set_visible(True)
ax.spines["bottom"].set_color("black")
ax.spines["bottom"].set_linewidth(1.5)
ax.tick_params(axis="both", which="major", labelsize=11, length=6, width=1.5, direction="out", colors="black")
ax.legend(fontsize=12, loc="best")

plt.tight_layout()
out = os.path.join(output_dir, "rmsf_fluctuation_glycans_vs_noglycans.png")
plt.savefig(out, dpi=800, bbox_inches="tight", facecolor="white")
plt.show()
print(f"Saved: {out}")
```
