# MUC2 No Glycans Analysis

- Plotting the RMSD and RMSF for all 3 runs


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-darkgrid')
```

## File Location

```python
# Time step
dt_ns = 0.12018  # ns per frame

# Output directory
output_dir = "/home/computer2/analysis_simulations/analysis/plots_muc2_noglycans"
os.makedirs(output_dir, exist_ok=True)

print("Output directory:", output_dir)
```

## Load Data - All 3 Runs

```python
# Run 1 - analysis_muc2_noglycans
df_rmsf_1 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans/rmsf.dat",
                         sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_rmsd_1 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans/rmsd.dat",
                          sep=r"\s+", skiprows=1, names=["Frame", "RMSD"])

# Run 2 - analysis_muc2_noglycans1
df_rmsf_2 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans1/rmsf_noglyc_1.dat",
                         sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_rmsd_2 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans1/rmsd_noglyc_1.dat",
                          sep=r"\s+", skiprows=1, names=["Frame", "RMSD"])

# Run 3 - analysis_muc2_noglycans3
df_rmsf_3 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans3/rmsf3_noglyc.dat",
                         sep=r"\s+", skiprows=1, names=["Residue", "RMSF"])
df_rmsd_3 = pd.read_csv("/home/computer2/analysis_simulations/analysis/analysis_muc2_noglycans3/rmsd_beginning_3.dat",
                          sep=r"\s+", skiprows=1, names=["Frame", "RMSD"])

# Convert frames to time (ns)
time_ns_1 = (df_rmsd_1["Frame"] - df_rmsd_1["Frame"].iloc[0]) * dt_ns
time_ns_2 = (df_rmsd_2["Frame"] - df_rmsd_2["Frame"].iloc[0]) * dt_ns
time_ns_3 = (df_rmsd_3["Frame"] - df_rmsd_3["Frame"].iloc[0]) * dt_ns

print("All 3 RMSF and RMSD datasets loaded")
print(f"Run 1 - RMSF: {df_rmsf_1.shape}, RMSD: {df_rmsd_1.shape}")
print(f"Run 2 - RMSF: {df_rmsf_2.shape}, RMSD: {df_rmsd_2.shape}")
print(f"Run 3 - RMSF: {df_rmsf_3.shape}, RMSD: {df_rmsd_3.shape}")
```

## RMSD - Individual Plots

```python
# RMSD - Run 1
fig, ax = plt.subplots(figsize=(14, 7), dpi=100)
ax.plot(time_ns_1, df_rmsd_1["RMSD"], linewidth=2.5, color="#2E86AB", alpha=0.85,
        marker="o", markersize=3, markevery=max(1, len(df_rmsd_1)//50), label="Run 1")
ax.fill_between(time_ns_1, df_rmsd_1["RMSD"], alpha=0.1, color="#2E86AB")
ax.set_xlabel("Time (ns)", fontsize=13, fontweight="bold")
ax.set_ylabel("RMSD (\u00c5)", fontsize=13, fontweight="bold")
ax.set_title("RMSD - No Glycans Run 1", fontsize=15, fontweight="bold", pad=20)
ax.set_ylim(bottom=0)
ax.grid(True, alpha=0.3, linestyle="--", linewidth=0.5)
ax.set_facecolor("white")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(True)
ax.spines["left"].set_color("black")
ax.spines["left"].set_linewidth(1.5)
ax.spines["bottom"].set_visible(True)
ax.spines["bottom"].set_color("black")
ax.spines["bottom"].set_linewidth(1.5)
ax.tick_params(axis="both", which="major", labelsize=11, length=6, width=1.5, direction="out", colors="black")
plt.tight_layout()
out = os.path.join(output_dir, "rmsd_noglycans_run1.png")
plt.savefig(out, dpi=800, bbox_inches="tight", facecolor="white")
plt.show()
print(f"Saved: {out}")
```

## RMSF Mean ± Std

```python
rmsf_all = pd.concat([
    df_rmsf_1.set_index("Residue")["RMSF"],
    df_rmsf_2.set_index("Residue")["RMSF"],
    df_rmsf_3.set_index("Residue")["RMSF"]
], axis=1, keys=["Run1", "Run2", "Run3"])

mean_rmsf = rmsf_all.mean(axis=1)
std_rmsf  = rmsf_all.std(axis=1)
residues  = mean_rmsf.index

fig, ax = plt.subplots(figsize=(14, 7), dpi=100)
ax.plot(residues, mean_rmsf, linewidth=2.5, color="#A23B72", label="Mean RMSF")
ax.fill_between(residues, mean_rmsf - std_rmsf, mean_rmsf + std_rmsf,
                alpha=0.3, color="#A23B72", label="\u00b1 Std")
ax.set_xlabel("Residue Number", fontsize=13, fontweight="bold")
ax.set_ylabel("RMSF (\u00c5)", fontsize=13, fontweight="bold")
ax.set_title("RMSF Mean \u00b1 Std - MUC2 No Glycans", fontsize=14, fontweight="bold", pad=15)
ax.set_xticks(np.arange(0, residues.max() + 100, 100))
ax.set_xlim(left=residues.min())
ax.set_ylim(bottom=0)
ax.legend(fontsize=12, loc="best")
ax.grid(True, alpha=0.3, linestyle="--", linewidth=0.5)
ax.set_facecolor("white")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(True)
ax.spines["left"].set_color("black")
ax.spines["left"].set_linewidth(1.5)
ax.spines["bottom"].set_visible(True)
ax.spines["bottom"].set_color("black")
ax.spines["bottom"].set_linewidth(1.5)
ax.tick_params(axis="both", which="major", labelsize=11, length=6, width=1.5, direction="out", colors="black")
plt.tight_layout()
out = os.path.join(output_dir, "rmsf_noglycans_mean_std.png")
plt.savefig(out, dpi=800, bbox_inches="tight", facecolor="white")
plt.show()
print(f"Saved: {out}")
```
